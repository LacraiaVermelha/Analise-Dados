```python
import random
import math
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/workspaces/Analise-Dados/Labs/Lab03SocialMediaData.csv', engine='python')
```

```python
print(f"Quantidade de linhas, colunas: {df.shape}")
print(f"Nome das colunas: {df.columns}")
print("Head")
print(df.head())
print("Tail")
print(df.tail())
```

```python
plt.hist(df['Platform'], color='red', orientation='horizontal')
plt.title('Distribuição de usuários entre plataformas')
plt.xlabel('Plataforma')
plt.ylabel('Quantidade de usuários')
plt.show()
```

```python
dailyMinutes = []
socialMediaNames = ['Reddit', 'Quora', 'WhatsApp', 'Instagram']
dailyMinutes.append(df.loc[df['Platform'] == 'Reddit', 'Daily Time Spent (min)'].mean())
dailyMinutes.append(df.loc[df['Platform'] == 'Quora', 'Daily Time Spent (min)'].mean())
dailyMinutes.append(df.loc[df['Platform'] == 'WhatsApp', 'Daily Time Spent (min)'].mean())
dailyMinutes.append(df.loc[df['Platform'] == 'Instagram', 'Daily Time Spent (min)'].mean())

plt.bar(socialMediaNames, dailyMinutes)
plt.title("Uso médio de cada media social por min/dia")
plt.show()
```
