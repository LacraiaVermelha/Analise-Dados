### Calculo de nota

```python
import random
import math
from matplotlib import pyplot as plt
```

```python
def mediaAritLista(lista1, lista2):
    ativMedias = []
    for i in range(len(lista1)):
        ativMedias.append((lista1[i] + lista2[i]) / 2)
    return ativMedias
```

```python
def mediaBimestre(ativ, prova):
    mediasBimestre = []
    for i in range(len(ativ)):
        mediasBimestre.append(ativ[i]*0.25 + prova[i]*0.75)
    return mediasBimestre
```

```python
def mediaFinal(media, np):
    mediasFinal = []
    for i in range(len(media)):
        mediasFinal.append(media[i] + np[i])
    return mediasFinal
```

```python
def calcMediaFinal(ativ1p1, ativ1p2, ativ2p1, ativ2p2, p1, p2, np):
    media1 = mediaBimestre(mediaAritLista(ativ1p1, ativ1p2), p1)
    media2 = mediaBimestre(mediaAritLista(ativ2p1, ativ2p2), p2)
    mediaPreNP = mediaAritLista(media1, media2)
    return mediaFinal(mediaPreNP, np)
```

```python
def roundList(list, precision):
    roundedList = []
    for i in range(len(list)):
        roundedList.append(round(list[i], precision))
    return roundedList
```

```python
def getMediana(lista):
    lista.sort()
    mediana = 0
    if (len(lista) % 2 == 0):
        mediana = (lista[math.floor(len(lista)/2)] + lista[math.floor(len(lista)/2 - 1)]) / 2
    else:
        mediana = lista[math.floor(len(lista)/2)]
    return mediana
```

```python
def rngNotas(lenght, min, max, divisor):
    notas = []
    for i in range(lenght):
        notas.append(random.randint(min, max) / divisor)
    return notas
```

```python
# Gerando Notas

ativ1Parte1 = rngNotas(20,0,100,10)
ativ1Parte2 = rngNotas(20,0,100,10)
ativ2Parte1 = rngNotas(20,0,100,10)
ativ2Parte2 = rngNotas(20,0,100,10)
p1 = rngNotas(20,0,100,10)
p2 = rngNotas(20,0,100,10)
np = rngNotas(20,0,5,10)

print("--- N1 ---")
print(ativ1Parte1)
print(ativ1Parte2)
print(p1)

print("--- N2 ---")
print(ativ2Parte1)
print(ativ2Parte2)
print(p2)

print("--- np ---")
print(np)

print("--- Média Final ---")
mediasFinais = calcMediaFinal(ativ1Parte1, ativ1Parte2, ativ2Parte1, ativ2Parte2, p1, p2, np)
print(roundList(mediasFinais, 2))
mediasFinais.sort()
print(mediasFinais)
print(getMediana(mediasFinais))
```
